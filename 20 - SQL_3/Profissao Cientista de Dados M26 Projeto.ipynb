Abaixo temos a configuração do ambiente SQL:

```python
import sqlite3
import pandas as pd
```

```python
df_transacoes = pd.read_csv("TB_TRANSACOES_PROJETO_ECOMM.csv", delimiter=';')
df_clientes = pd.read_csv("TB_CLIENTES_PROJETO_ECOMM.csv", delimiter=';')
```

```python
conn = sqlite3.connect('projeto.db')
# Carregar o DataFrame no banco de dados SQLite - criando tb_transacoes e tb_clientes
df_transacoes.to_sql('tb_transacoes', conn, index=False, if_exists='replace')
df_clientes.to_sql('tb_clientes', conn, index=False, if_exists='replace')
```

```python
# Função para executar consultas SQL e retornar o resultado como um DataFrame
def run_query(query):
    return pd.read_sql_query(query, conn)
```

```python
query = """
SELECT *
FROM tb_clientes
"""
result_df = run_query(query)
print(result_df)
```

```python
query = """
SELECT *
FROM tb_transacoes
ORDER BY id_client
limit 10
"""
result_df = run_query(query)
print(result_df)
```

```python
query = """
SELECT
A.*,
COUNT(B.id_client) AS Total_Purchases,
SUM(B.Price) AS Total_Spent
FROM tb_clientes AS A
INNER JOIN tb_transacoes AS B
ON A.Id_client = B.id_client
GROUP BY A.Id_client
ORDER BY Total_Purchases DESC
"""
result_df = run_query(query)
print(result_df)
```

Justifique a escolha do JOIN:

Foi utilizado um INNER JOIN para relacionar as tabelas 'tb_clientes' e 'tb_transacoes' por meio do "id_client". Essa escolha foi feita porque a análise tem como foco clientes que realizaram compras. Não foi usado um RIGHT JOIN pois existem registros na tabela 'tb_transacoes' sem cliente associado, que são desconsiderados pelo INNER JOIN. E um LEFT JOIN não foi utilizado porque retornaria clientes sem compras registradas, o que não é o objetivo da análise.
As colunas "Total_Purchases" e "Total_Spent" foram criadas para registar as informações de cada cliente em uma única linha, representando respectivamente a quantidade de compras realizadas e o valor total gasto. E essas escolhas realizadas, reduzem o tamanho do 'arquivo.csv' gerado, evitando que um mesmo cliente apareça várias vezes devido a múltiplas transações, facilitando na visualização e compreensão do DataFrame.

Exportando o arquivo como CSV:
